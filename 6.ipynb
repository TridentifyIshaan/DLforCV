# Variables

This notebook focuses particularly on variables.

````python
x = tf.constant([1,2])
x_var = tf.Variable(x)
print(x_var)
````

````python
x_var = tf.Variable(x, name='var1')
print(x_var)
````

````python
# assign - Assigns a new value to the variable.

v = tf.Variable(1.)
v.assign(2.)
````

````python
# assign_add - Adds a value to this variable.
# Similary for assign_sub #[1,2] -> [0,0]

x_var.assign_add([3,4])
````

# Types of Tensors

This notebook focuses particularly on these types of tensors:-
- Ragged Tensors
- Sparse Tensors
- String Tensors

````python
tensor_two_d = tf.constant([
    [1,2,0],
    [3,5,-1],
    [1,5,6],
    [2,3,8]
])
tensor_two_d.shape
````

````python
# Ragged Tensor - Without it, you can't keep your column values flexible, it will show error.

tensor_two_d =[[1,2,0],
                [3,],
                [1,5,6,5,6],
                [2,3]]

print(tf.ragged.constant(tensor_two_d).shape) # None as it is not fixed
tf.ragged.constant(tensor_two_d)
````

````python
# tf.ragged.boolean_mask - Applies a boolean mask to data without flattening the mask dimensions.

# Aliases for True & False so data and mask line up.
T, F = (True, False)

tf.ragged.boolean_mask(  # Mask a 2D Tensor.
    data=[[1, 2, 3], [4, 5, 6], [7, 8, 9]],
    mask=[[T, F, T], [F, F, F], [T, F, F]]
).to_list()
````

````python
tf.ragged.boolean_mask(  # Mask a 2D RaggedTensor.
    tf.ragged.constant([[1, 2, 3], [4], [5, 6]]),
    tf.ragged.constant([[F, F, T], [F], [T, T]])
).to_list()
````

````python
tf.ragged.boolean_mask(  # Mask rows of a 2D RaggedTensor.
    tf.ragged.constant([[1, 2, 3], [4], [5, 6]]),
    tf.ragged.constant([True, False, True])
).to_list()
````

Whole documentation on ragged tensor - https://www.tensorflow.org/api_docs/python/tf/RaggedTensor

````python
# from_row_lengths - Creates a RaggedTensor with rows partitioned by row_lengths.
print(tf.RaggedTensor.from_row_lengths(
    values=[3, 1, 4, 1, 5, 9, 2, 6],
    row_lengths=[4, 0, 3, 1, 0])
)
````

````python
# from_row_limits - Creates a RaggedTensor with rows partitioned by row_limits.
print(tf.RaggedTensor.from_row_limits(
    values=[3, 1, 4, 1, 5, 9, 2, 6],
    row_limits=[4, 4, 7, 8, 8])
)
````

````python
# from_row_splits - Creates a RaggedTensor with rows partitioned by row_splits.

print(tf.RaggedTensor.from_row_splits(
    values=[3, 1, 4, 1, 5, 9, 2, 6],
    row_splits=[0, 4, 4, 7, 8, 8]) # splits: 0 to 4, 4 to 4, 4 to 7, 7 to 8, 8 to 8
)
````

````python
# tf.sparse.SparseTensor - Represents a sparse tensor.

t = tf.sparse.SparseTensor(indices=[[1, 1], [3, 4]], values=[11, 56], dense_shape=[5, 6])
print(t)
````

````python
tf.sparse.to_dense(t)
````

````python
# String Tensor
s = tf.constant(["Hello", "I am", "a string"])
print(s)
````

````python
# join
tf.strings.join(s, separator =" ")
````

````python
# tf.strings.length - String lengths of input.
strings = tf.constant(['Hello','TensorFlow', '\U0001F642'])
tf.strings.length(strings).numpy() # default counts bytes

tf.strings.length(strings, unit="UTF8_CHAR").numpy()
````

````python
# tf.strings.lower - Converts all uppercase characters into their respective lowercase replacements.
tf.strings.lower("CamelCase string and ALL CAPS")
````

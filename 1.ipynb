# Tensors

## Types of Arrays / Tensors

- 0-D Array - 8               - NO SHAPE
- 1-D Array - [2 0 -3]        - Shape = (3,)
- 2-D array / matrix

          - Shape = (4,3)
- 3-D array - Combination of different matrices         - Shape = (4,2,3)

````python
import tensorflow as tf
import numpy as np
import pandas as pd
````

````python
# 0-D, 1-D, 2-D, 3-D

tensor_zero_d = tf.constant(4)
print(tensor_zero_d)
print()

tensor_one_d = tf.constant([2,0.,-3], dtype = tf.float32)
print(tensor_one_d)
print()

tensor_two_d = tf.constant([[1,2,0], [3,5,-1], [1,5,6], [2,3,8]])
print(tensor_two_d)
print()

tensor_three_d = tf.constant([ [[1,2,0], [3,5,-1]], [[10,2,0],[1,0,2]], [[5,8,0],[2,7,0]], [[2,1,9], [4,-3,32]] ])
print(tensor_three_d)
print()
````

````python
print(tensor_two_d[0:3, 0:2]) # indexing in multidimensonal tensor
print(tensor_two_d[..., 0:2]) # ... can be replaced by :
````

````python
# Shape
print(tensor_zero_d.shape, tensor_one_d.shape, tensor_two_d.shape, tensor_three_d.shape, sep="\n")
# Dimension
print(tensor_zero_d.ndim, tensor_one_d.ndim, tensor_two_d.ndim, tensor_three_d.ndim, sep="\n")
````

````python
# Shape
print(tf.shape(tensor_three_d)) # To recieve the array's shape as aninput and tell the shape of that shape array.
````

````python
#4-D

tensor_four_d = tf.constant([ [[[1,2,0], [3,5,1]], [[10,2,0],[1,0,2]], [[5,8,0],[2,7,0]], [[2,1,9], [4,-3,32]]],
    [[[13,26,0], [3,5,-12]], [[10,2,0],[1,0,23]], [[5,8,0],[2,73,0]], [[2,1,9], [4,-30,32]]],
    [[[103,26,0], [3,50,-12]], [[100,2,0],[1,0,23]], [[5,28,0],[2,3,0]], [[22,1,9], [44,-320,32]]],
 ])
print(tensor_four_d)
print()
````

````python
# when you have typecaste a float to integer, you have to use cast function, otherwise it will be an error
casted_tensor_one_d = tf.cast(tensor_one_d, dtype = tf.int16)
print(casted_tensor_one_d)
````

````python
# Boolean Tensor

tensor_bool = tf.constant([True, True, False])
print(tensor_bool)

# String Tensor

tensor_string = tf.constant(["Hello World!", "Namaste Duniya!"])
print(tensor_string)
````

````python
# NumPy Array Conversion into a Tensor
np_array = np.array([1,2,4])
np_array
````

````python
converted_tensor = tf.convert_to_tensor(np_array)
print(converted_tensor)
````

````python
# eye - construct an identity matrix, or a batch of matrices
eye_tensor = tf.eye(
    num_rows = 5,
    num_columns=3,
    batch_shape=[2,4],
    dtype=tf.dtypes.float32,
    name=None
)
print(eye_tensor)
````

````python
# fill - Creates a tensor filled with a scalar value.
fill_tensor = tf.fill(
    # dims, value, name=None, layout=None
    [3,4], 5
)
print(fill_tensor)
````

````python
# ones - Creates a tensor with all elements set to one (1).

# NOTE - SIMILARLY, zeros

ones_tensor = tf.ones( 
    # shape, dtype=tf.dtypes.float32, name=None, layout=None
    [5,3,2], dtype=tf.dtypes.float32
)
print(ones_tensor)
````

````python
# ones_like - Creates a tensor of all ones that has the same shape as the input.

# NOTE - SIMILARLY, zeros_like

ones_like_tensor = tf.ones_like(
    # input, dtype=None, name=None, layout=None
    [[12,1,3],[5,7,2]]
)
print(ones_like_tensor)
````

````python
# rank - Returns the rank of a tensor.

tf.rank(tensor_four_d)
````

````python
# size - Returns the size of a tensor.

tf.size(tensor_four_d, out_type=tf.float32) # product of rows & columns

# out_type - receive input in some other datatype and return the size in the datatype set in out_type
````

````python
# random.normal - Outputs random values from a normal distribution.

random_normal_tensor = tf.random.normal(
    # shape, mean=0.0, stddev=1.0, dtype=tf.dtypes.float32, seed=None, name=None
    [3,2], mean=6.0, stddev=2.5, dtype=tf.dtypes.float32, seed=None, name=None
)
print(random_normal_tensor)
````

Normal Probablity Tester - https://www.acsu.buffalo.edu/~adamcunn/probability/normal.html

````python
# random.uniform - Outputs random values from a uniform distribution.

random_uniform_tensor = tf.random.uniform(
    # shape, minval=0, maxval=None, dtype=tf.dtypes.float32, seed=None, name=None
    [5,], minval=1, maxval=4, dtype=tf.dtypes.float32, seed=None, name=None # By default, maxval = 1
)
print(random_uniform_tensor)
````

````python
# seed - A Python integer. Used in combination with tf.random.set_seed to create a reproducible sequence of tensors across multiple calls.

tf.random.set_seed(5)
print(tf.random.uniform(shape=[3,], maxval=5, dtype=tf.int32, seed=10))
print(tf.random.uniform(shape=[3,], maxval=5, dtype=tf.int32, seed=10))
print(tf.random.uniform(shape=[3,], maxval=5, dtype=tf.int32, seed=10))
print(tf.random.uniform(shape=[3,], maxval=5, dtype=tf.int32, seed=10))
````
